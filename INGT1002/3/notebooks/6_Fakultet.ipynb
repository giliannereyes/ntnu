### Om fakultetsoperasjonen

En sentral matteoperasjon i Statistikk er [*Fakultet*](https://en.wikipedia.org/wiki/Factorial), som matematisk skrives med utropstegn $n!$, og utføres på *heltall*. Formelen for utregning av fakultet er som følger:
$$n! = \prod_{k=1}^{n} k = 1\cdot2\cdot3\cdot4\cdot\ldots\cdot n$$

Du har kanskje ikke sett $\prod$ brukt på denne måten i en matteformel før, men som du ser betyr det kumulativt produkt fra nedre heltall til øvre heltall. I programmering løses dette typisk med en oppsamlingsløkke.

<a id="oppg_8a"></a>
## a)

Fullfør en funksjon som regner ut fakultet med bruk av for-løkke. Du skal _ikke_ ta i bruk noen funksjoner fra `math`-modulen i denne oppgaven.

```python
def fakultet(n):
    ### BEGIN SOLUTION
    if n < 0:
        return float("nan")
    n_fakultet = 1
    for i in range(1, n+1):
        n_fakultet*=i
    ### END SOLUTION
    return n_fakultet
```

__Test av utregning:__

Noe som er verdt å bemerke seg er at fakultetsoperasjonen ikke er definert for negative tall, og laveste verdi $n$ med en definerbar $n!$ er null, hvor $0! = 1$. Om funksjonen `fakultet(n)` fra del**a)** skal brukes videre så kan det være lurt at den klarer å gi et passende output for tilfeller der $n < 0$. En mulighet da er å returnere `nan` som er en forkortelse for **n**ot **a** **n**umber, som ofte brukes når resultatet av en beregning er et tall som ikke eksisterer. Merk at verdien fremdeles behandles av python som om det var et tall, selv symbolene `nan` minner mer om tekst.

**Eksempel:**

```python
ugyldig_tall = float("NaN") 
print(ugyldig_tall)
```

## b)
Oppdater funksjonen `fakultet(n)` i del**a)** slik at den returnerer `nan` dersom tallet $n$ er et negativt tall. For $n\geq 0$ skal den fortsatt returnere $n!$ som vanlig.

```python
### BEGIN HIDDEN TESTS
from unittest.mock import patch
from autofeedback import FunctionTests
from random import sample, randint
from numpy import isnan

def fakultet_solution(n):
    if n < 0:
        return float("nan")
    n_fakultet = 1
    for i in range(1, n+1):
        n_fakultet*=i
    return n_fakultet

grading_results = FunctionTests(fakultet_solution)
try:
    grading_results.add_test_func(fakultet)
except Exception as e:
    grading_results.log.append("Could not run tests, "+e.args[0])
else:
    x_vals =  sample(list(range(1, 15)), 1) + [0]
    for x in x_vals:
        try:
            # Check for calls to "abs" function
            with patch("math.factorial") as mock_factorial:
                _ = fakultet(x)
        except Exception as e:
            pass
        finally:
            if mock_factorial.called:
                grading_results.add_result(False, f"Built-in function 'math.factorial()' called during the execution of function call 'abs_value({x})'")
            else:
                grading_results.test_return_value(x)
    x_negative = randint(-10, -1)
    return_value = fakultet(x_negative)
    if isnan(return_value):
        grading_results.add_result(True, f"function call 'factorial({x_negative})' returned 'nan' which is expected", wgt = 10)
    else:
        grading_results.add_result(False, f"function call 'factorial({x_negative})' did not return 'nan' as expected", wgt = 10)
round(grading_results.get_results()*5, 2)
### END HIDDEN TESTS
```
